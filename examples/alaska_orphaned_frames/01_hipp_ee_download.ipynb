```python
# ------------------------
# Imports
# ------------------------

import hipp
from getpass import getpass
import os
```

```python
# ------------------------
# Authenticate EROS API
# ------------------------

apiKey = hipp.dataquery.EE_login(input(), getpass())
```

```python
# ------------------------
# Set query parameters
# ------------------------

bounds = (-179, 72, -130, 50)      # Bounds for Alaska
#bounds = (0, 0, 0, 0)               # Bounds for orphan projects
ULLON, ULLAT, LRLON, LRLAT = bounds
ee_agency_code = '4'
ee_project_name = 'AK590'
year = 1959
month = 1
day = 1

output_directory    = os.path.join('.\\data', 'raw_data', 'EEFrames', ee_project_name)
ee_query_max_results   = 50000 #do more than 1 because you may get a calibration file
ee_query_label = 'test_download' #try to use the same one for consistent behavior from the EE API
```

```python
# ------------------------
# Run query
# ------------------------

ee_results_df = hipp.dataquery.EE_pre_select_images(
    apiKey,
    xmin = LRLON,
    ymin = LRLAT,
    xmax = ULLON,
    ymax = ULLAT,
    startDate = f"{year}-{month}-{day}",
    endDate = f"{year+1}-{month}-{day}",
    maxResults   = ee_query_max_results
)
```

```python
# ------------------------
# Show query as DataFrame
# ------------------------

ee_results_df = ee_results_df[ee_results_df['agency'] == ee_agency_code]
ee_results_df = ee_results_df[ee_results_df['project'] == ee_project_name]
ee_results_df
```

```python
# ------------------------
# Download images
# ------------------------

images_directory, calibration_reports_directory, pixel_pitch = hipp.dataquery.EE_download_images_to_disk(
    apiKey,
    ee_results_df['entityId'].tolist(),
    ee_query_label,
    output_directory
)
```
